# Dependencias

```python
from typing import cast
from copy import deepcopy

from graphs.plot_graph import plot_graph
from graphs.graph_loader import load_graph_from_file
from graphs.undirected_graph import UndirectedGraph
from graphs.directed_graph import DirectedGraph
```

# Ejemplo

Nombre del archivo CSV a utilizar.

```python
filename = "csv\\ejemplo_6.csv"
```

Se carga el grafo del archivo CSV como dirigido y no dirigido. Se castean las variables para no confundir al editor de código.

```python
directed_graph = cast(
  DirectedGraph,
  load_graph_from_file(filename, True)
)
directed_graph_copy = deepcopy(directed_graph)
undirected_graph = cast(
  UndirectedGraph,
  load_graph_from_file(filename, False)
)
undirected_graph_copy = deepcopy(undirected_graph)
```

Visualización del grafo dirigido original

```python
plot_graph(directed_graph)
```

Visualización del grafo no dirigido original

```python
plot_graph(undirected_graph)
```

```python
directed_graph.get_vertices()
```
